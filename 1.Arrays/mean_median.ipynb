Average of all elements in an array


```python
def findMean(arr):
    sum = arr[0]

    for i in range(1, len(arr)):
        sum += arr[i]

    average = sum // len(arr)
    return average


arr = [1, 2, 3, 4, 5]
findMean(arr)
```
