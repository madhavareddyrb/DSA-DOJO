1636. Sort Array by Increasing Frequency

Given an array of integers nums, sort the array in increasing order based on the frequency of the values. If multiple values have the same frequency, sort them in decreasing order.

Return the sorted array.

Example 2:

Input: nums = [2,3,1,3,2]
Output: [1,3,3,2,2]
Explanation: '2' and '3' both have a frequency of 2, so they are sorted in decreasing order.

##### STEPS:

1. Every thing is based on frequncy of elemenst so we must know freq of all elemenst using dict
2. find values with freq
3. find keys with more frequncy and order of values if both matches
4. print array with frequncy less to high in decsending orderr


```python
def sortarrIncreasingFreq(arr):
    freq = {}
    for i in arr:
        if i in freq:
            freq[i] += 1
        else:
            freq[i] = 1
    print(freq)
    # new_arr = []
    # for key, value in freq.items():
    #     new_arr.append(value * str(key).split(","))
    arr = sorted(arr, key=lambda x: (freq[x], -x))

    return arr


arr = [1, 4, 4, 3, 2, 2, 3, 3, 4, 6]
sortarrIncreasingFreq(arr)
```

```python
def sum_arr(arr):
    sum = 0
    for i in range(len(arr)):
        sum += arr[i]
        arr[i] = sum
    return arr

arr = [1, 2, 3, 4, 5]
print(sum_arr(arr))
```

189. Rotate Array

Given an integer array nums, rotate the array to the right by k steps, where k is non-negative.

Example 1:

Input: nums = [1,2,3,4,5,6,7], k = 3
Output: [5,6,7,1,2,3,4]

```python
nums = [1,2,3,4,5,6,7]
k = 3
for i in range(1,k+1):
    pop_v = nums.pop()
    nums.insert(0, pop_v)
    
print(nums)
```

1. Two Sum

Given an array of integers nums and an integer target, return indices of the two numbers such that they add up to target.

You may assume that each input would have exactly one solution, and you may not use the same element twice.

You can return the answer in any order.

Example 1:

Input: nums = [2,7,11,15], target = 9
Output: [0,1]
Explanation: Because nums[0] + nums[1] == 9, we return [0, 1].


```python
def two_sum(arr, target):
    for i in range(len(arr)):
        for j in range(len(arr)-1, i,-1):
            if arr[i] + arr[j]== target:
            	return i,j


arr = [3,2,4]
target = 6
print(two_sum(arr, target))

arr2 =[ 2, 7, 11, 15]
target = 9
print(two_sum(arr2, target))
```

```python
def removeDuplicates(nums) :
    if len(nums) == 0:
        return 0
    uniq = 0
    for i in range(1, len(nums)):
        if nums[i] != nums[uniq]:
            uniq += 1
            nums[uniq] = nums[i]

    return uniq + 1

nums = [1,1,2,2,2,3,3]
removeDuplicates(nums)
```
