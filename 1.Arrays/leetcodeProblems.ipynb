1636. Sort Array by Increasing Frequency

Given an array of integers nums, sort the array in increasing order based on the frequency of the values. If multiple values have the same frequency, sort them in decreasing order.

Return the sorted array.

Example 2:

Input: nums = [2,3,1,3,2]
Output: [1,3,3,2,2]
Explanation: '2' and '3' both have a frequency of 2, so they are sorted in decreasing order.

##### STEPS:

1. Every thing is based on frequncy of elemenst so we must know freq of all elemenst using dict
2. find values with freq 
3. find keys with more frequncy and order of values if both matches
4. print array with frequncy less to high in decsending orderr


```python
def sortarrIncreasingFreq(arr):
    freq = {}
    for i in arr:
        if i in freq:
            freq[i] += 1
        else:
            freq[i] = 1
    print(freq)
    # new_arr = []
    # for key, value in freq.items():
    #     new_arr.append(value * str(key).split(","))
    arr = sorted(arr, key=lambda x: (freq[x], -x))

    return arr


arr = [1, 4, 4, 3, 2, 2, 3, 3, 4, 6]
sortarrIncreasingFreq(arr)
```
