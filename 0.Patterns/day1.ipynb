Square of side 'N' n * n


```python
n = 3
for i in range(n):
    for j in range(n):
        print("*", end = "")
    print()
```

Hollow Square of side 'N'

Example:

Input: 3
Output: ['***', '* *', '***']
 
Input: 5
Output: ['*****', '*   *', '*   *', '*   *', '*****']

```python
def hollowSquare(n):
    result = []
    for i in range(n):
        row = ""
        for j in range(n):
            if i == n - 1 or i == 0 or j == 0 or j == n - 1:
                row += "*"
            else:
                row += " "
        result.append(row)

    return result

n = 5
hollowSquare(n)
```
